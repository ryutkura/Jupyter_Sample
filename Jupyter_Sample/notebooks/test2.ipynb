````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 平均 (mu) と標準偏差 (sigma)
mu = 1
sigma = 1

# xの範囲
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)

# 正規分布の確率密度関数 (PDF) の計算
pdf = norm.pdf(x, mu, sigma)

# グラフの描画
plt.figure(figsize=(10, 6))
plt.plot(x, pdf, label='Gaussian Distribution (μ=1, σ=1)', color='blue')
plt.title('Gaussian Distribution (Normal Distribution)')
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 平均と標準偏差の設定
mu1, sigma1 = 1, 1
mu2, sigma2 = 1, 2

# xの範囲
x = np.linspace(mu1 - 4*sigma1, mu1 + 4*sigma1, 1000)

# 正規分布の確率密度関数 (PDF) の計算
pdf1 = norm.pdf(x, mu1, sigma1)
pdf2 = norm.pdf(x, mu2, sigma2)

# グラフの描画
plt.figure(figsize=(10, 6))
plt.plot(x, pdf1, label='Gaussian Distribution (μ=1, σ=1)', color='blue')
plt.plot(x, pdf2, label='Normal Distribution (μ=1, σ=2)', color='red')
plt.title('Gaussian and Normal Distributions')
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
plt.show()
````
